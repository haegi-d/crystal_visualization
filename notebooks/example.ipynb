# crystal_visualization — example notebook

This notebook walks through the full pipeline from structure file to publication figure.
Swap `backend` between `'matplotlib'`, `'plotly'`, and `'blender'` to switch rendering quality.

```python
from pathlib import Path
from crystal_visualization import CrystalRenderer

cr = CrystalRenderer()
```

## 1. Load structure
Accepts XYZ, CIF, POSCAR, or any format supported by ASE.

```python
cr.load_structure("../data/Er_Y2O3.xyz")
```

## 2. Select defect and neighbor shell
Use the `plotly` backend first to explore the structure interactively.
Hover over atoms to see species, index, and distance from the defect.

```python
(
    cr
    .select_defect(species="Er", index=0)
    .set_neighbor_cutoff(angstrom=5.0)
    .set_camera(view="isometric")
    .set_style(style="default")
    .render(output=Path("../output/preview.html"), backend="plotly", trim={"O": "Er"})
)
```

## 3. Draft figure with matplotlib
Quick 2D projection — no Blender required. Good for checking colors and layout.

```python
cr.set_style(style="nature").render(
    output=Path("../output/draft.pdf"),
    backend="matplotlib",
)
```

## 4. Final publication render with Blender
Requires Blender to be installed. See DESIGN.md §2 for setup.
Pass `save_blend=True` to keep the `.blend` file for manual inspection.

```python
cr.render(
    output=Path("../output/figure.pdf"),
    backend="blender",
    save_blend=True,
)
```
